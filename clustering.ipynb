```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# New dataset
data = {
    "Product": [
        "Nutrela soya chunks", "Healthy and Tasty Mustard Oil", "MAGGI® Pasta Cheesy Tomato",
        "MAGGI® 2-Minute Noodles", "AMUL Cheese", "Rice Idli", "Thaker Diary Paneer",
        "Snack Sweet Corn", "Red Cow Ghee", "Mordern Burger Supreme",
        "Keventer Chicken Sausage", "Keventer Chicken Nuggets", "Keventer Chicken Strips"
    ],
    "Total Fat (g)": [0.82, 100, 1.9, 15.7, 25, 2.3, 25, 2, 99.7, 3.23, 8, 11.6, 9.9],
    "Protein (g)": [53.19, 0, 7.7, 8, 20, 13, 20, 3.2, 0, 9.5, 12.3, 14.7, 15.7],
    "Carbs (g)": [33.5, 0, 69.8, 63.5, 1.5, 67, 2.5, 14.4, 0, 49.7, 8.6, 14.2, 16.9]
}

df = pd.DataFrame(data)

# Selecting features for clustering
features = df[["Total Fat (g)", "Protein (g)", "Carbs (g)"]]

# Standardizing the features
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

# Define clustering algorithms
clustering_algorithms = {
    'KMeans': KMeans(n_clusters=3, random_state=0),
    'DBSCAN': DBSCAN(eps=1, min_samples=2),
    'Agglomerative': AgglomerativeClustering(n_clusters=3)
}

# Color palette (used Codeium for AI code Suggestions)
palette = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

# Plot and print cluster components for each algorithm
for name, algorithm in clustering_algorithms.items():
    # Apply clustering
    df['Cluster'] = algorithm.fit_predict(features_scaled)

    # 3D Plotting the clusters
    fig = plt.figure(figsize=(12, 8))
    ax = fig.add_subplot(111, projection='3d')

    for cluster in df['Cluster'].unique():
        cluster_data = df[df['Cluster'] == cluster]
        color = palette[cluster % len(palette)] if cluster != -1 else 'gray'  # gray for noise
        label = f'Cluster {cluster}' if cluster != -1 else 'Noise'
        ax.scatter(cluster_data['Total Fat (g)'], cluster_data['Protein (g)'], cluster_data['Carbs (g)'],
                   color=color, label=label, s=100)

    # Adding product labels to the points with a slight offset for clarity
    for i, product in enumerate(df['Product']):
        ax.text(df['Total Fat (g)'][i] + 2, df['Protein (g)'][i] + 2, df['Carbs (g)'][i] + 2,
                product, fontsize=9)

    ax.set_title(f"Clustering of Products Based on Fat, Protein, and Carbs ({name})")
    ax.set_xlabel("Total Fat (g)")
    ax.set_ylabel("Protein (g)")
    ax.set_zlabel("Carbs (g)")
    ax.legend(title="Product Clusters", loc='upper left', bbox_to_anchor=(1.05, 1))
    plt.tight_layout()
    plt.show()

    # USED CHATGPT for generating the following display code
    # Display cluster components
    print(f"\n{name} Cluster Components:")
    for cluster in sorted(df['Cluster'].unique()):
        cluster_name = f"Cluster {cluster}" if cluster != -1 else "Noise"
        products_in_cluster = df[df['Cluster'] == cluster]['Product'].tolist()
        print(f"{cluster_name}: {', '.join(products_in_cluster)}")
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Data for clustering
data = {
    "Product": [
        "MAGGI® 2-Minute Noodles", "AMUL Cheese", "Nutrela soya chunks",
        "Healthy and Tasty Mustard Oil", "Red Cow Ghee", "Rice Idli",
        "Thaker Diary Paneer", "MAGGI® Pasta Cheesy Tomato", "Snack Sweet Corn",
        "Mordern Burger Supreme", "Keventer Chicken Sausage",
        "Keventer Chicken Nuggets", "Keventer Chicken Strips"
    ],
    "Total Energy (kcal)": [427, 311, 354.14, 900, 897, 351, 310, 327, 79, 266, 155.8, 220, 219.8],
    "Protein (g)": [8, 20, 53.19, 0, 0, 13, 20, 7.7, 3.2, 9.5, 12.3, 14.7, 15.7],
    "Carbs (g)": [63.5, 1.5, 33.5, 0, 0, 67, 2.5, 69.8, 14.4, 49.7, 8.6, 14.2, 16.9],
    "Sugars (g)": [2.2, 1.5, 7.18, 0, 0, 0.5, 2.5, 6.8, 3.5, 3.1, 0, 0.4, 0.1],
    "Fiber (g)": [3.6, 0, 0, 0, 0, 6.5, 0, 0, 4.5, 0, 0, 0, 0],
    "Total Fat (g)": [15.7, 25, 0.82, 100, 99.7, 2.3, 25, 1.9, 2, 3.23, 8, 11.6, 9.9],
    "Saturated Fat (g)": [6.8, 16, 0.19, 99, 65, 1.1, 17.6, 1.5, 0.3, 2.45, 3.1, 2.8, 1.6],
    "Sodium (g)": [1172.3, 1200, 8, 0, 0, 900, 12, 712.9, 400, 479, 0, 0, 0]
}

df = pd.DataFrame(data)

# Selecting relevant features for clustering
features = df[["Total Energy (kcal)", "Protein (g)", "Carbs (g)", "Sugars (g)", "Fiber (g)", "Total Fat (g)", "Saturated Fat (g)", "Sodium (g)"]]

# Standardizing the features
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

# Apply PCA for dimensionality reduction (used Codeium for AI code Suggestions)
pca = PCA(n_components=2)
principal_components = pca.fit_transform(features_scaled)
df['PC1'] = principal_components[:, 0]
df['PC2'] = principal_components[:, 1]

# Define clustering algorithms
clustering_algorithms = {
    'KMeans': KMeans(n_clusters=3, random_state=0),
    'DBSCAN': DBSCAN(eps=1, min_samples=2),
    'Agglomerative': AgglomerativeClustering(n_clusters=3)
}

# Color palette (used Codeium for AI code Suggestions)
palette = ['#1f77b4', '#ff7f0e', '#2ca02c']

# Plot for each algorithm and print clusters
for name, algorithm in clustering_algorithms.items():
    df['Cluster'] = algorithm.fit_predict(features_scaled)

    # Plotting the clusters with distinctive legends and improved labels
    plt.figure(figsize=(12, 8))
    for cluster in df['Cluster'].unique():
        cluster_data = df[df['Cluster'] == cluster]
        color = palette[cluster % len(palette)] if cluster != -1 else 'gray'
        label = f'Cluster {cluster}' if cluster != -1 else 'Noise'
        plt.scatter(cluster_data['PC1'], cluster_data['PC2'], color=color, label=label, s=100)

    # Adding product labels to the points with adjust_text for overlap management
    texts = [plt.text(df['PC1'][i], df['PC2'][i], product, fontsize=9) for i, product in enumerate(df['Product'])]

    plt.title(f"Clustering of Products Based on Nutritional Information ({name})")
    plt.xlabel("Principal Component 1")
    plt.ylabel("Principal Component 2")
    plt.legend(title="Product Clusters")
    plt.grid(True)
    plt.show()

    # USED CHATGPT for generating the following display code
    # Printing cluster components
    print(f"\n{name} Cluster Components:")
    for cluster in sorted(df['Cluster'].unique()):
        cluster_name = f"Cluster {cluster}" if cluster != -1 else "Noise"
        products_in_cluster = df[df['Cluster'] == cluster]['Product'].tolist()
        print(f"{cluster_name}: {', '.join(products_in_cluster)}")
```
